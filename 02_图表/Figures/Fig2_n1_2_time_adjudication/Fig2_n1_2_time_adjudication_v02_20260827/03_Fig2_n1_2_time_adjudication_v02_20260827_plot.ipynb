# PRL P6 T32/T64/T128 时间裁决 Figure 2

本 Notebook 只读取版本包内冻结的真实 P6 输出。结论边界：固定 D0/E0/F150 基线在 T64→T128 下通过全部 P6 时间门；P5 历史失败保持归档，不外推为空间收敛或 EFE 机制。

```python
from pathlib import Path
import importlib.util
from IPython.display import Image, Markdown, display

PREFIX = "Fig2_n1_2_time_adjudication_v02_20260827"
REVISION_DIR = Path.cwd().resolve()
STYLE_SOURCE = "用户批准当前阶段自由排版的科研审阅图"
FIGURE_SIZE_IN = (18.2, 10.8)
DPI = 220
HELPER = REVISION_DIR / f"03d_{PREFIX}_plot_helper.py"
OUTPUT_PNG = REVISION_DIR / f"04_{PREFIX}.png"
OUTPUT_SVG = REVISION_DIR / f"05_{PREFIX}.svg"
print(f"当前版本目录: {REVISION_DIR}")
```

## 真实数据绘图与导出

辅助代码读取版本包内 global、field、closure、dense-waveform 和 peak-field 文件，生成 PNG/SVG；不创建或修改任何模型结果。

```python
if not HELPER.is_file():
    raise FileNotFoundError(HELPER)
spec = importlib.util.spec_from_file_location("figure2_time_adjudication_helper", HELPER)
module = importlib.util.module_from_spec(spec)
if spec.loader is None:
    raise RuntimeError("无法加载绘图辅助代码")
spec.loader.exec_module(module)
png_path, svg_path = module.build_figure(REVISION_DIR)
print(png_path.name, svg_path.name)
```

## 最终导出预览

```python
if not OUTPUT_PNG.is_file() or OUTPUT_PNG.stat().st_size == 0:
    raise FileNotFoundError(OUTPUT_PNG)
if not OUTPUT_SVG.is_file() or OUTPUT_SVG.stat().st_size == 0:
    raise FileNotFoundError(OUTPUT_SVG)
display(Markdown(f"### 当前预览：{OUTPUT_PNG.name}"))
display(Image(filename=str(OUTPUT_PNG)))
```
